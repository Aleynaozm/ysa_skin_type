```python
# Ana veri klasörlerini tanımla (Adım 3 için kullanılacak)
BASE_DIR = '/content/skin_dataset'
```

```python
# ============================================
# 3. VERİ SETİ YAPISINI DÜZENLEME (HATA DÜZELTME)
# ============================================
print("\n🔧 Klasör yapısı kontrol ediliyor ve veri yolu düzeltiliyor...")

# Veri setinin gerçek yapılarını (Dry, Normal, Oily klasörlerini) bul
skin_folders = []
for root, dirs, files in os.walk(BASE_DIR):
    for d in dirs:
        # Sadece sınıf isimlerini içeren klasörleri bul
        if d.lower() in ['dry', 'oily', 'normal']:
            skin_folders.append(os.path.join(root, d))

print(f"✅ Bulunan sınıf klasörleri: {len(skin_folders)}")
for folder in skin_folders:
    num_images = len([f for f in os.listdir(folder) if f.lower().endswith(('.jpg', '.jpeg', '.png'))])
    print(f"  - {folder}: {num_images} görüntü")

if len(skin_folders) != 3:
    print("❌ Kritik Hata: 3 adet sınıf klasörü bulunamadı. Lütfen indirilen dosya yapısını manuel kontrol edin.")
else:
    print("✅ Klasör yapısı başarılı!")
```

```python
# ============================================
# 3. VERİ SETİ YAPISINI DÜZENLEME (VAR OLAN YAPILARI KULLANMA)
# ============================================
print("\n🔧 Veri yapısı düzeltiliyor: Önceden bölünmüş veriler bulundu!")

# Önceki hatalı boş klasörleri temizle
try:
    shutil.rmtree(f'{BASE_DIR}/train')
    shutil.rmtree(f'{BASE_DIR}/validation')
    shutil.rmtree(f'{BASE_DIR}/test')
except:
    pass

# Var olan TRAIN/VALIDATION/TEST klasörlerini içeren ana dizini tanımla
BASE_DIR_PRE_SPLIT = f'{BASE_DIR}/Oily-Dry-Skin-Types'

# Yolları TensorFlow jeneratörleri için güncelleyelim:
TRAIN_DIR = f'{BASE_DIR_PRE_SPLIT}/train'
VAL_DIR = f'{BASE_DIR_PRE_SPLIT}/valid' # Veri setinde 'validation' yerine 'valid' kullanılmış
TEST_DIR = f'{BASE_DIR_PRE_SPLIT}/test'

print(f"✅ Yeni TRAIN yolu: {TRAIN_DIR}")
print(f"✅ Yeni VALIDATION yolu: {VAL_DIR}")
print(f"✅ Yeni TEST yolu: {TEST_DIR}")
```

```python
# Kontrol
print("\n📊 Final veri seti DAĞILIMI (Var Olan Yapı):")

all_data = []
for split, path in [('Train', TRAIN_DIR), ('Validation', VAL_DIR), ('Test', TEST_DIR)]:
    total = 0
    row_data = {'Split': split}
    # Sınıf isimlerini al
    class_names_found = [d for d in os.listdir(path) if os.path.isdir(os.path.join(path, d))]

    for class_name in class_names_found:
        class_path = f"{path}/{class_name}"
        count = len(os.listdir(class_path))
        total += count
        row_data[class_name.capitalize()] = count
    row_data['Total'] = total
    all_data.append(row_data)

distribution_df = pd.DataFrame(all_data)
print(distribution_df.to_string(index=False))
print("✅ Adım 3 tamamlandı. Veri yolları jeneratörler için hazır.")
```

```python
# ============================================
# 4. HİPERPARAMETRELER
# ============================================
print("\n⚙️ Hiperparametreler ayarlanıyor...")

IMG_SIZE = (224, 224)
BATCH_SIZE = 32
EPOCHS = 50
NUM_CLASSES = 3
CLASS_NAMES = ['dry', 'normal', 'oily']
LEARNING_RATE = 0.0001  # Tüm modeller için optimize edilmiş öğrenme oranı

print(f"✅ Görüntü boyutu: {IMG_SIZE}")
print(f"✅ Batch size: {BATCH_SIZE}")
print(f"✅ Epoch sayısı: {EPOCHS}")
print(f"✅ Sınıf sayısı: {NUM_CLASSES}")
print(f"✅ Öğrenme oranı: {LEARNING_RATE}")
```

```python

# ============================================
# 5. VERİ ARTIRMA VE GENERATOR'LAR
# ============================================
print("\n🎨 5. Data augmentation (Regularization) ve jeneratörler yapılandırılıyor...")
# Slaytlarda Veri Artırma, Regularization tekniği olarak belirtilmiş [cite: 420, 422]

train_datagen = ImageDataGenerator(
    rescale=1./255, rotation_range=20, width_shift_range=0.2, height_shift_range=0.2,
    shear_range=0.15, zoom_range=0.15, horizontal_flip=True, fill_mode='nearest',
    brightness_range=[0.8, 1.2]
)
val_test_datagen = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_directory(
    TRAIN_DIR, target_size=IMG_SIZE, batch_size=BATCH_SIZE, class_mode='categorical', shuffle=True, seed=42
)
val_generator = val_test_datagen.flow_from_directory(
    VAL_DIR, target_size=IMG_SIZE, batch_size=BATCH_SIZE, class_mode='categorical', shuffle=False
)
test_generator = val_test_datagen.flow_from_directory(
    TEST_DIR, target_size=IMG_SIZE, batch_size=BATCH_SIZE, class_mode='categorical', shuffle=False
)
```

```python
# 6. VERİ GÖRSELLEŞTİRME (Sınıf Dağılımı)
# ============================================
print("\n📊 6. Sınıf dağılımı kontrol ediliyor...")

class_counts = pd.DataFrame({
    'Train': [sum(train_generator.classes == i) for i in range(NUM_CLASSES)],
    'Validation': [sum(val_generator.classes == i) for i in range(NUM_CLASSES)],
    'Test': [sum(test_generator.classes == i) for i in range(NUM_CLASSES)]
}, index=CLASS_NAMES)

print(class_counts)

fig, ax = plt.subplots(figsize=(12, 6))
class_counts.plot(kind='bar', ax=ax, color=['#FF6B6B', '#4ECDC4', '#45B7D1'], edgecolor='black', linewidth=2)
plt.title('Sınıf Dağılımı (Train/Validation/Test)', fontsize=16, fontweight='bold')
plt.xticks(rotation=0)
plt.show()
```

```python
# 7. MODEL 1: BASELINE CNN (Derste Öğrenilen Teknikler)
# ============================================
print("\n" + "="*80)
print("🤖 7. MODEL 1: BASELINE CNN (ReLU, Softmax, Adam, Dropout)")
print("="*80)

def create_baseline_cnn():
    model = models.Sequential([
        # Conv Block 1
        layers.Conv2D(32, (3, 3), activation='relu', padding='same', input_shape=(224, 224, 3)), # ReLU
        layers.BatchNormalization(),
        layers.MaxPooling2D((2, 2)),
        layers.Dropout(0.25), # Dropout

        # Conv Block 2
        layers.Conv2D(64, (3, 3), activation='relu', padding='same'),
        layers.BatchNormalization(),
        layers.MaxPooling2D((2, 2)),
        layers.Dropout(0.25),

        # Fully Connected
        layers.Flatten(),
        layers.Dense(512, activation='relu'),
        layers.BatchNormalization(),
        layers.Dropout(0.5),
        layers.Dense(256, activation='relu'),
        layers.BatchNormalization(),
        layers.Dropout(0.4),
        layers.Dense(NUM_CLASSES, activation='softmax') # Softmax
    ])
    return model

model_baseline = create_baseline_cnn()

# Optimize edilmiş öğrenme oranı kullanılıyor (0.0001)
model_baseline.compile(
    optimizer=Adam(learning_rate=LEARNING_RATE), # Adam Optimizer
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
print("✅ Baseline CNN modeli tanımlandı ve derlendi.")
```

```python
# 8. MODEL 2: VGG16 (TRANSFER LEARNING)
# ============================================
print("\n" + "="*80)
print("🤖 8. MODEL 2: VGG16 (Feature Extraction)")
print("="*80)

def create_vgg16_model():
    base_model = VGG16(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
    base_model.trainable = False
    model = models.Sequential([
        base_model,
        layers.GlobalAveragePooling2D(),
        layers.Dense(512, activation='relu'), layers.BatchNormalization(), layers.Dropout(0.5),
        layers.Dense(256, activation='relu'), layers.BatchNormalization(), layers.Dropout(0.3),
        layers.Dense(NUM_CLASSES, activation='softmax')
    ])
    return model

model_vgg16 = create_vgg16_model()
model_vgg16.compile(optimizer=Adam(learning_rate=LEARNING_RATE), loss='categorical_crossentropy', metrics=['accuracy'])
print("✅ VGG16 modeli tanımlandı ve derlendi.")
```

```python
# 9. MODEL 3: MOBILENETV2 (Fine-Tuning)
# ============================================
print("\n" + "="*80)
print("🤖 9. MODEL 3: MOBILENETV2 (Fine-Tuning)")
print("="*80)

def create_mobilenet_model():
    base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
    base_model.trainable = True
    for layer in base_model.layers[:100]:
        layer.trainable = False
    model = models.Sequential([
        base_model,
        layers.GlobalAveragePooling2D(),
        layers.Dense(256, activation='relu'), layers.BatchNormalization(), layers.Dropout(0.4),
        layers.Dense(128, activation='relu'), layers.BatchNormalization(), layers.Dropout(0.3),
        layers.Dense(NUM_CLASSES, activation='softmax')
    ])
    return model

model_mobilenet = create_mobilenet_model()
model_mobilenet.compile(optimizer=Adam(learning_rate=LEARNING_RATE), loss='categorical_crossentropy', metrics=['accuracy'])
print("✅ MobileNetV2 modeli tanımlandı ve derlendi.")

```

```python

# 10. MODEL 4: RESNET50 (Fine-Tuning)
# ============================================
print("\n" + "="*80)
print("🤖 10. MODEL 4: RESNET50 (Fine-Tuning)")
print("="*80)

def create_resnet_model():
    base_model = ResNet50(weights='imagenet', include_top=False, input_shape=(224, 224, 3))
    base_model.trainable = True
    for layer in base_model.layers[:-20]:
        layer.trainable = False
    model = models.Sequential([
        base_model,
        layers.GlobalAveragePooling2D(),
        layers.Dense(512, activation='relu'), layers.BatchNormalization(), layers.Dropout(0.5),
        layers.Dense(NUM_CLASSES, activation='softmax')
    ])
    return model

model_resnet = create_resnet_model()
model_resnet.compile(optimizer=Adam(learning_rate=LEARNING_RATE), loss='categorical_crossentropy', metrics=['accuracy'])
print("✅ ResNet50 modeli tanımlandı ve derlendi.")
```

```python
# ============================================
# 11. TÜM MODELLERİN TOPLANMASI
# ============================================
print("\n" + "="*80)
print("📋 11. TÜM MODELLER HAZIR - EĞİTİM İÇİN HAZIR")
print("="*80)

# Modelleri toplu sözlüğe ekle (Eğitim Adımı için)
models_dict = {
    'Baseline_CNN': model_baseline,
    'VGG16': model_vgg16,
    'MobileNetV2': model_mobilenet,
    'ResNet50': model_resnet
}

print(f"✅ Toplam {len(models_dict)} model hazır:")
for name in models_dict.keys():
    print(f"  - {name}")

print("\n💡 Sonraki adım: Model eğitimi ve değerlendirme")
```

```python
# ============================================
# 12. CALLBACK'LER (Early Stopping, Learning Rate Reduction, Model Checkpoint)
# ============================================
print("\n" + "="*80)
print("⚙️ 12. CALLBACK'LER AYARLANIYOR")
print("="*80)

# Early Stopping: Overfitting'i önlemek için
early_stopping = EarlyStopping(
    monitor='val_loss',
    patience=10,
    restore_best_weights=True,
    verbose=1
)

# Learning Rate Reduction: Öğrenme oranını dinamik olarak azalt
reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=5,
    min_lr=1e-7,
    verbose=1
)

# Model Checkpoint: En iyi modeli kaydet
checkpoint_dir = '/content/checkpoints'
os.makedirs(checkpoint_dir, exist_ok=True)

callbacks_list = [early_stopping, reduce_lr]

print("✅ Callback'ler hazır:")
print("  - Early Stopping: val_loss'u izler, 10 epoch sabır")
print("  - Reduce LR on Plateau: val_loss durursa LR'yi yarıya indirir")
print("  - Model Checkpoint: En iyi modeli kaydeder")
```

```python
# ============================================
# 15. TEK GÖRÜNTÜ TAHMİN FONKSİYONU (ARAYÜZ İÇİN)
# ============================================
print("\n" + "="*80)
print("🖼️ 15. TAHMİN FONKSİYONU HAZIRLANIYOR")
print("="*80)

from PIL import Image

def predict_skin_type(image_path, model):
    """
    Tek bir görüntü için cilt tipi tahmini yapar
    
    Args:
        image_path: Görüntü dosya yolu
        model: Eğitilmiş model
    
    Returns:
        dict: Tahmin sonuçları (sınıf, olasılıklar)
    """
    # Görüntüyü yükle ve ön işle
    img = Image.open(image_path)
    img = img.convert('RGB')
    img = img.resize(IMG_SIZE)
    img_array = np.array(img) / 255.0
    img_array = np.expand_dims(img_array, axis=0)
    
    # Tahmin yap
    predictions = model.predict(img_array, verbose=0)
    predicted_class_idx = np.argmax(predictions[0])
    predicted_class = CLASS_NAMES[predicted_class_idx]
    confidence = predictions[0][predicted_class_idx] * 100
    
    # Tüm sınıflar için olasılıklar
    probabilities = {
        CLASS_NAMES[i]: float(predictions[0][i] * 100) 
        for i in range(len(CLASS_NAMES))
    }
    
    return {
        'predicted_class': predicted_class,
        'confidence': confidence,
        'probabilities': probabilities
    }

print("✅ Tahmin fonksiyonu hazır!")
print("\n💡 Kullanım örneği:")
print("   result = predict_skin_type('path/to/image.jpg', model_to_train)")
print("   print(f'Tahmin: {result[\"predicted_class\"]} ({result[\"confidence\"]:.2f}%)')")
```
